# Leitura e tratamento dos dados

DATA COM: É dada que se pode comprar para receber o pagamento referido na linha

DY: É o valor do dividendo sobre o valor da ação

```python
planilha = pd.read_csv(r"C:\Users\luis.nascimento\Documents\Análises Financeiras\(Cru) FIIs rendimentos por data (2021-2025).csv")
planilha
```

```python
planilha.info()
```

```python
planilha["VALOR"] = planilha["VALOR"].apply(lambda x: x.replace("R$ ", ""))
# planilha["VALOR"] = planilha["VALOR"].apply(lambda x: float(x.split('.')[0] + '.' + x.split('.')[1][:2]))
planilha["DATA COM"] =  planilha["DATA COM"].apply(lambda x: pd.to_datetime(x.replace('/', '-'), format='%d-%m-%Y'))
planilha["DATA PAGAMENTO"] =  planilha["DATA PAGAMENTO"].apply(lambda x: pd.to_datetime(x.replace('/', '-'), format='%d-%m-%Y'))
planilha["DY"] =  planilha["DY"].apply(lambda x: x.replace('%', ''))
planilha
```

```python
lista_ano = []
for i in planilha.index:
    lista_ano.append(planilha.loc[i, 'DATA PAGAMENTO'].year)
planilha["ANO"] = lista_ano
```

```python
planilha.sort_values(['ATIVO','DATA PAGAMENTO'])
```

```python
lista_ativos = []
lista_qtd_2023 = []
lista_qtd_2024 = []
lista_qtd_2025 = []

lista_total_2023 = []
lista_total_2024 = []
lista_total_2025 = []

for i in planilha.index:
    ativo = planilha.loc[i, 'ATIVO']
    if ativo not in lista_ativos: 
        lista_ativos.append(ativo)
        df_aux = planilha.loc[planilha['ATIVO'] == ativo]
        # df_aux.sort_values("ANO_MES")
        lista_qtd_2023.append(len(pd.unique(df_aux.loc[df_aux["ANO"] == 2023]["ANO_MES"])))
        lista_qtd_2024.append(len(pd.unique(df_aux.loc[df_aux["ANO"] == 2024]["ANO_MES"])))
        lista_qtd_2025.append(len(pd.unique(df_aux.loc[df_aux["ANO"] == 2025]["ANO_MES"])))

        lista_total_2023.append(sum(df_aux.loc[df_aux["ANO"] == 2023]["VALOR"]))
        lista_total_2024.append(sum(df_aux.loc[df_aux["ANO"] == 2024]["VALOR"]))
        lista_total_2025.append(sum(df_aux.loc[df_aux["ANO"] == 2025]["VALOR"]))
    
df_ativos = pd.DataFrame({"Ativo": lista_ativos,
                         "Quantidade em 2023": lista_qtd_2023,
                         "Quantidade em 2024": lista_qtd_2024,
                         "Quantidade em 2025": lista_qtd_2025,
                         "Total pago em 2023": lista_total_2023,
                         "Total pago em 2024": lista_total_2024,
                         "Total pago em 2025": lista_total_2025})


df_ativos
# ativo_ano.append(df_aux["DATA PAGAMENTO"].year)
# ativo_ano.append(df_aux["DATA PAGAMENTO"].year)
```
