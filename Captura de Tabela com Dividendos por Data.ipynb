# Captura tabela

```python
tabela_dividendo = wd.find_element(By.XPATH, '//*[@id="datePag"]/div/div/div[1]/div[2]/div/table')
```

```python
tabela_dados = []

linhas = tabela_dividendo.find_elements(By.TAG_NAME, "tr")
for linha in linhas:
    colunas = linha.find_elements(By.TAG_NAME, "td")
    if not colunas:
        colunas = linha.find_elements(By.TAG_NAME, "th")
    tabela_dados.append([coluna.text for coluna in colunas])

print(len(tabela_dados))
```

```python
df = pd.DataFrame(tabela_dados[1:], columns=tabela_dados[0])
df
```

```python
df.to_csv(r".\Dados Capturados\(Cru) FIIs rendimentos por data (2021-2025).csv", index=False)
```
