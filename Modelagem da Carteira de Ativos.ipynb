# Bibliotecas

```python
import pandas as pd
```

# Ativos da Carteira

```python
df_ativos = pd.read_csv('Dados Modelados/Ativos.csv')
lista_ativos_selecionados = [
    # Ações
    "BBSE3",
	"PETR4",
	"TAEE11",
	"VALE3",

    # FIIs
    "HGLG11",
    "HTMX11",
    "KNCR11",
    "XMLP11",
    "XPCI11"

]

lista_index = []
for i in df_ativos.index:
    if df_ativos.loc[i, "ATIVO"] in lista_ativos_selecionados:
        lista_index.append(i)

df_ativos = df_ativos.loc[lista_index, :]
df_ativos.index = range(0, len(df_ativos))
df_ativos['Tipo'] = "Ação"
df_ativos
```

```python
df_fiis = pd.read_csv('Dados Modelados/FIIs.csv')

lista_ativos_selecionados = [
    # Ações
    "BBSE3",
	"PETR4",
	"TAEE11",
	"VALE3",

    # FIIs
    "HGLG11",
    "HTMX11",
    "KNCR11",
    "XPML11",
    "XPCI11"

]

lista_index = []
for i in df_fiis.index:
    if df_fiis.loc[i, "ATIVO"] in lista_ativos_selecionados:
        lista_index.append(i)

df_fiis = df_fiis.loc[lista_index, :]
df_fiis.index = range(0, len(df_fiis))
df_fiis['Tipo'] = "FII"
df_fiis
```

```python
df_carteira_ativos = pd.concat([df_ativos, df_fiis])
df_carteira_ativos.index = range(0, len(df_carteira_ativos))
df_carteira_ativos
```

```python
acoes = df_carteira_ativos.loc[df_carteira_ativos['Tipo'] == "Ação"]['ATIVO'].values
acoes.sort()
print(acoes)

fundos = df_carteira_ativos.loc[df_carteira_ativos['Tipo'] == "FII"]['ATIVO'].values
fundos.sort()
print(fundos)
```

```python
balanco = {
    'BBSE3': 3,
    'PETR4': 30,
    'TAEE11': 3,
    'VALE3': 3,

    'HGLG11': 3,
    'HTMX11': 3, 
    'KNCR11': 3, 
    'XPCI11': 3,
    'XPML11': 3
}

balanco_ativos = []
for i in df_carteira_ativos.index:
    balanco_ativos.append(
        balanco[df_carteira_ativos.loc[[i], "ATIVO"].values[0]]
    )

df_carteira_ativos["Balanço de Cotas"] = balanco_ativos
df_carteira_ativos
```

```python
df_carteira_ativos.to_csv('Dados Modelados/carteira_ativos.csv', index=False)
```
