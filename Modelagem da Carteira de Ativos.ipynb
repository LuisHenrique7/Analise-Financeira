# Ativos da Carteira

```python
df_ativos = pd.read_csv('Dados Modelados/Ativos.csv')
lista_ativos_selecionados = [
    # Ações
    "BBSE3",
	"PETR4",
	"TAEE11",
	"VALE3",

    # FIIs
    "HGLG11",
    "HTMX11",
    "KNCR11",
    "XMLP11",
    "XPCI11"

]

lista_index = []
for i in df_ativos.index:
    if df_ativos.loc[i, "ATIVO"] in lista_ativos_selecionados:
        lista_index.append(i)

df_ativos = df_ativos.loc[lista_index, :]
df_ativos.index = range(0, len(df_ativos))
df_ativos['Tipo'] = "Ação"
df_ativos
```

```python
df_fiis = pd.read_csv('Dados Modelados/FIIs.csv')

lista_ativos_selecionados = [
    # Ações
    "BBSE3",
	"PETR4",
	"TAEE11",
	"VALE3",

    # FIIs
    "HGLG11",
    "HTMX11",
    "KNCR11",
    "XPML11",
    "XPCI11"

]

lista_index = []
for i in df_fiis.index:
    if df_fiis.loc[i, "ATIVO"] in lista_ativos_selecionados:
        lista_index.append(i)

df_fiis = df_fiis.loc[lista_index, :]
df_fiis.index = range(0, len(df_fiis))
df_fiis['Tipo'] = "FII"
df_fiis
```

```python
df_carteira_ativos = pd.concat([df_ativos, df_fiis])
df_carteira_ativos.index = range(0, len(df_carteira_ativos))
df_carteira_ativos
```

# Monta o Calendário dos Ativos da Carteira

## Histórico

```python
df_ativos = pd.read_csv('Dados Modelados/historico_Ativos.csv')
df_ativos = df_ativos.loc[df_ativos["Ativo"].isin(lista_ativos_selecionados)]
# df_ativos.index = range(0, len(df_ativos))
# df_ativos['Tipo Ativo'] = "Ação"
df_ativos
```

```python
df_fiis = pd.read_csv('Dados Modelados/historico_Fii.csv')
df_fiis = df_fiis.loc[df_fiis["Ativo"].isin(lista_ativos_selecionados)]
# df_fiis.index = range(0, len(df_fiis))
# df_fiis['Tipo Ativo'] = "FII"
df_fiis
```

```python
df_historico = pd.concat([df_ativos, df_fiis])
df_historico = df_historico.filter(["Ativo", "Data", "close"])
df_historico.rename(columns={"close": "Valor"}, inplace=True)
df_historico["Natureza"] = ["Histórico de Preço"] * len(df_historico)
df_historico["Data"] = pd.to_datetime(df_historico["Data"], format="%d-%m-%Y")
df_historico.index = range(0, len(df_historico))
df_historico
```

```python
df_historico.info()
```

## Dividendos

```python
df_dividendos = pd.concat(
    [
        pd.read_csv('Dados Modelados/Dividendos pagos por data (2021-2025).csv'), 
        pd.read_csv('Dados Modelados/FIIs rendimentos por data (2021-2025).csv')
    ])
df_dividendos = df_dividendos.loc[df_dividendos['ATIVO'].isin(lista_ativos_selecionados)]
df_dividendos.rename(columns={"ATIVO": "Ativo", "VALOR": "Valor", "TIPO": "Tipo Receita", "DATA PAGAMENTO": "Data"}, inplace=True)
df_dividendos = df_dividendos.filter(["Ativo", "Data", "Valor", "DATA COM", "Tipo Receita", "DY"])
df_dividendos["Natureza"] = ["Dividendos"] * len(df_dividendos)
df_dividendos["Valor"] = df_dividendos["Valor"].apply(lambda x: str(x).replace(',', '.'))
df_dividendos["Valor"] = df_dividendos["Valor"].apply(
    lambda x: float(str(x).split('.')[0] + '.' + str(x).split('.')[1][:2])
)

df_dividendos["Data"] = pd.to_datetime(df_dividendos["Data"], format="%Y-%m-%d")
df_dividendos["DATA COM"] = pd.to_datetime(df_dividendos["DATA COM"], format="%Y-%m-%d")
df_dividendos
```

```python
df_dividendos.info()
```

## Aquisição

```python
df_aquisicao = pd.read_csv("Exemplo Aquisição.csv")
df_aquisicao.rename(columns={"Preço de Aquisição": "Valor", "Data de Aquisição": "Data"}, inplace=True)
df_aquisicao = df_aquisicao.filter(["Ativo", "Data", "Valor", "Número de Cotas", "Dono"])
df_aquisicao["Natureza"] = ["Aquisição"] * len(df_aquisicao)
df_aquisicao['Valor'] = df_aquisicao['Valor'].str.replace(',', '.').astype(float)
df_aquisicao["Data"] = pd.to_datetime(df_aquisicao["Data"], format="%d-%m-%Y")
df_aquisicao
```

```python
df_aquisicao.info()
```

## Une os dados

```python
df_historico_carteira = pd.concat([df_historico, df_dividendos])
df_historico_carteira = pd.concat([df_historico_carteira, df_aquisicao])
df_historico_carteira
```

```python
df_historico_carteira.sort_values(by="Data", inplace=True)
df_historico_carteira.index = range(0, len(df_historico_carteira))
df_historico_carteira
```

```python
df_historico_carteira.loc[df_historico_carteira["Natureza"] == "Aquisição"]
```

```python
a = df_historico_carteira.loc[df_historico_carteira["Data"] >= '2024-04-01']
a
```

```python
lista_valor_investido = []
lista_preco_unitario = []
lista_preco_total = []
lista_dividendos_total = []
lista_dividendos_total_d = []

lista_valor_investido_acumulado_distinto = []
lista_preco_total_acumulado_distinto = []
# lista_valor_investido_acumulado = []
# lista_preco_total_acumulado = []
lista_dividendos_acumulados = []

dicionario_acumulados_distintos = {
    "Heitor": [0, 0, 0],
    "Luis": [0, 0, 0]
}

# valor_investido_acumulado = 0
# preco_total_acumulado = 0
dividendos_acumulados = 0

for i in a.index:
    ativo = a.loc[i, "Ativo"]
    data = a.loc[i, "Data"]
    valor = a.loc[i, "Valor"]
    dono = a.loc[i, "Dono"]

    valor_investido = 0
    preco_total = 0
    valor_dividendos = 0

    if a.loc[i, "Natureza"] == "Aquisição":
        preco = df_historico.loc[(df_historico["Ativo"] == ativo) & (df_historico["Data"] == data)]["Valor"].values[0]
        num_cotas = a.loc[i, 'Número de Cotas']
        valor_investido = num_cotas * valor
        preco_total = preco * num_cotas

        lista_valor_investido.append(valor_investido)
        lista_preco_unitario.append(preco)
        lista_preco_total.append(preco_total)
    else:
        lista_valor_investido.append(None)
        lista_preco_unitario.append(None)
        lista_preco_total.append(None)

    if a.loc[i, "Natureza"] == "Dividendos":
        data_com = a.loc[i, "DATA COM"]
        num_cotas = df_aquisicao.loc[(df_aquisicao["Ativo"] == ativo) & (df_aquisicao["Data"] <= data_com)]["Número de Cotas"].sum()
        valor_dividendos = valor * num_cotas

        lista_dividendos_total.append(valor_dividendos)

        num_cotas = df_aquisicao.loc[
            (df_aquisicao["Ativo"] == ativo) & (df_aquisicao["Data"] <= data_com) & (df_aquisicao["Dono"] == dono)
        ]["Número de Cotas"].sum()
        valor_dividendos_d = valor * num_cotas
        lista_dividendos_total_d.append(valor_dividendos_d)
    else:
        lista_dividendos_total.append(None)
        lista_dividendos_total_d.append(None)

    if dono == "Luis" or dono == "Heitor":
        dicionario_acumulados_distintos[dono][0] += valor_investido
        lista_valor_investido_acumulado_distinto.append(dicionario_acumulados_distintos[dono][0])
        dicionario_acumulados_distintos[dono][1] += preco_total
        lista_preco_total_acumulado_distinto.append(dicionario_acumulados_distintos[dono][1])
    else:
        lista_valor_investido_acumulado_distinto.append(None)
        lista_preco_total_acumulado_distinto.append(None)

    # valor_investido_acumulado += valor_investido
    # preco_total_acumulado += preco_total
    dividendos_acumulados += valor_dividendos

    # lista_valor_investido_acumulado.append(valor_investido_acumulado)
    # lista_preco_total_acumulado.append(preco_total_acumulado)
    lista_dividendos_acumulados.append(dividendos_acumulados)


a["Valor Investido"] = lista_valor_investido
a["Valor Investido Acumulado"] = lista_valor_investido_acumulado_distinto
# a["Valor Investido Acumulado"] = lista_valor_investido_acumulado
a["Preço Unitário"] = lista_preco_unitario
a["Preço Total"] = lista_preco_total
a["Preço Total Acumulado"] = lista_preco_total_acumulado_distinto
# a["Preço Total Acumulado"] = lista_preco_total_acumulado
a["Dividendos Totais D"] = lista_dividendos_total_d
a["Dividendos Totais"] = lista_dividendos_total
a["Dividendos Acumulados"] = lista_dividendos_acumulados
a
```
