# Bibliotecas

```python
import pandas as pd
import yahooquery as yq
```

# Leitura e tratamento dos dados

DATA COM: É dada que se pode comprar para receber o pagamento referido na linha

DY: É o valor do dividendo sobre o valor da ação

```python
planilha = pd.read_csv(r".\Dados Capturados\(Cru) Dividendos pagos por data (2021-2025).csv")
planilha
```

```python
planilha.info()
```

```python
planilha["VALOR"] = planilha["VALOR"].apply(lambda x: x.replace("R$ ", "").replace(",", '.'))
planilha["VALOR"] = planilha["VALOR"].apply(lambda x: float(x.split('.')[0] + '.' + x.split('.')[1][:2]))
planilha["DATA COM"] =  planilha["DATA COM"].apply(lambda x: pd.to_datetime(x.replace('/', '-'), format='%d-%m-%Y'))
planilha["DATA PAGAMENTO"] =  planilha["DATA PAGAMENTO"].apply(lambda x: pd.to_datetime(x.replace('/', '-'), format='%d-%m-%Y'))
planilha["DY"] =  planilha["DY"].apply(lambda x: float(x.replace(',', '.').replace('%', '')))
planilha
```

```python
planilha.info()
```

```python
pd.unique(planilha.loc[:, "ATIVO"])
```

```python
planilha.to_csv(r".\Dados Modelados\Dividendos pagos por data (2021-2025).csv", index=False)
```

# Criando histórico de preços

```python
ativos = pd.unique(planilha.loc[:, "ATIVO"]).tolist()
```

```python
lista_tickers = []
for ativo in ativos:
    lista_tickers.append((ativo,yq.Ticker(ativo+'.SA')))
```

```python
idx = 0
lista_precos_historicos = []
for nome, ticker in lista_tickers:
    lista_precos_historicos.append(ticker.history(period='5y',interval='1mo'))
    lista_precos_historicos[idx].assign(ativo=nome)
    idx += 1
```

```python
df_historico_ativos = pd.concat(lista_precos_historicos)
```

```python
df_historico_ativos['Ativo'] = df_historico_ativos.index.get_level_values(0)
df_historico_ativos['Data'] = df_historico_ativos.index.get_level_values(1)
```

```python
for idx, line in df_historico_ativos.iterrows():
    df_historico_ativos.loc[idx,'Ativo'] = df_historico_ativos.loc[idx,'Ativo'].split('.')[0]
```

```python
df_historico_ativos['Data'] = pd.to_datetime(df_historico_ativos['Data'], format='%Y-%m-%d')
```

```python
df_historico_ativos['Data'] = df_historico_ativos['Data'].apply(lambda x: x.strftime('%d-%m-%Y'))
```

```python
df_historico_ativos.to_csv('Dados Modelados/historico_ativos.csv', index=False)
```

# Criando setor, insdustria e recomendação

```python
idx = 0
ativos_se_in_re = pd.DataFrame()
for nome, ticker in lista_tickers:
    ativos_se_in_re.loc[idx, 'ATIVO'] = nome
    ativos_se_in_re.loc[idx, 'SETOR'] = ticker.asset_profile[nome+'.SA']['industry']
    ativos_se_in_re.loc[idx, 'INDUSTRIA'] = ticker.asset_profile[nome+'.SA']['sector']
    ativos_se_in_re.loc[idx, 'DATA'] = '11-09-2025'
    ativos_se_in_re.loc[idx, 'RECOMENDACAO'] = ticker.financial_data[nome+'.SA']['recommendationKey']
    idx += 1
```

```python
ativos.to_csv('Dados Modelados/Ativos.csv', index=False)
```

# Análise dos dados

```python
planilha.sort_values('DATA PAGAMENTO')
```

```python
lista_ano = []
for i in planilha.index:
    lista_ano.append(planilha.loc[i, 'DATA PAGAMENTO'].year)
planilha["ANO"] = lista_ano
```

```python
planilha.sort_values(['ATIVO','DATA PAGAMENTO'])
```

```python
lista_ativos = []
lista_qtd_2023 = []
lista_qtd_2024 = []
lista_qtd_2025 = []

lista_total_2023 = []
lista_total_2024 = []
lista_total_2025 = []

for i in planilha.index:
    ativo = planilha.loc[i, 'ATIVO']
    if ativo not in lista_ativos: 
        lista_ativos.append(ativo)
        df_aux = planilha.loc[planilha['ATIVO'] == ativo]
        
        lista_qtd_2023.append(len(pd.unique(df_aux.loc[df_aux["ANO"] == 2023]["ANO_MES"])))
        lista_qtd_2024.append(len(pd.unique(df_aux.loc[df_aux["ANO"] == 2024]["ANO_MES"])))
        lista_qtd_2025.append(len(pd.unique(df_aux.loc[df_aux["ANO"] == 2025]["ANO_MES"])))

        lista_total_2023.append(sum(df_aux.loc[df_aux["ANO"] == 2023]["VALOR"]))
        lista_total_2024.append(sum(df_aux.loc[df_aux["ANO"] == 2024]["VALOR"]))
        lista_total_2025.append(sum(df_aux.loc[df_aux["ANO"] == 2025]["VALOR"]))
    
df_ativos = pd.DataFrame({"Ativo": lista_ativos,
                         "Quantidade em 2023": lista_qtd_2023,
                         "Quantidade em 2024": lista_qtd_2024,
                         "Quantidade em 2025": lista_qtd_2025,
                         "Total pago em 2023": lista_total_2023,
                         "Total pago em 2024": lista_total_2024,
                         "Total pago em 2025": lista_total_2025})


df_ativos
```

```python
df_ativos.to_csv(r".\Ativos.csv" , index=False)
```
